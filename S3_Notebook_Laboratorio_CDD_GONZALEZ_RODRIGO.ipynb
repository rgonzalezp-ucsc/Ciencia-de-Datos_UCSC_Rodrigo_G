### ✏️ Actividad 1.2 — Media contra mediana

Para tu variable numérica principal, compara media, mediana y sesgo. Decide cuál de las dos reportarías en un informe y justifícalo.

```python
# 1.2 - MEDIA CONTRA MEDIANA

variable = df['cnt']

media = variable.mean()
mediana = variable.median()
sesgo = variable.skew()

print('Variable: Demanda total (cnt)')
print('Media:', round(media, 2))
print('Mediana:', round(mediana, 2))
print('Asimetría:', round(sesgo, 3))

print('\nDiferencia entre media y mediana:',
      round(abs(media - mediana), 2))
```

_Justificación:_Se recomienda utilizar la media (4.504,35 alquileres), ya que la diferencia con la mediana (4.548) es pequeña y la asimetría es prácticamente nula (-0,047). Esto indica que la distribución es relativamente simétrica y que la media representa adecuadamente la demanda diaria.

### ✏️ Actividad 2.1 — Comportamiento por grupo

Agrupa por tu variable categórica más importante y compara una métrica numérica entre grupos. Usa `groupby` con al menos tres estadísticos (`count`, `median`, `std`).

```python
# ACTIVIDAD 2.1 - COMPORTAMIENTO POR GRUPO

resumen = df.groupby('season')['cnt'].agg(
    count='count',
    media='mean',
    mediana='median',
    desviacion='std'
).round(2)

resumen
```

### ✏️ Actividad 2.2 — Matriz de correlación

Calcula y grafica la matriz de correlación de tus variables numéricas. Identifica los dos pares con mayor correlación.

```python
corr = df.select_dtypes(include='number').corr()

fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(corr, cmap='RdBu_r', vmin=-1, vmax=1)

ax.set_xticks(range(len(corr)))
ax.set_xticklabels(corr.columns, rotation=90)
ax.set_yticks(range(len(corr)))
ax.set_yticklabels(corr.columns)

for i in range(len(corr)):
    for j in range(len(corr)):
        ax.text(j, i, f'{corr.iloc[i, j]:.2f}',
                ha='center', va='center', fontsize=8)

plt.colorbar(im)
plt.title('Matriz de correlación')
plt.tight_layout()
plt.show()
```

### ✏️ Actividad 2.3 — El par más correlacionado, bajo sospecha

Toma el par con mayor correlación y **no te quedes con el número**. Haz el recorrido del Ejercicio 4:

1. Grafica el diagrama de dispersión. ¿La relación es lineal, o el coeficiente engaña?
2. ¿Alguna de las dos variables depende del **tamaño** de la unidad de análisis? Si es así, **normalízalas** y vuelve a calcular.
3. Piensa qué **tercera variable** podría estar causando ambas, y calcula la correlación parcial.
4. Si tienes una variable de agrupación, repite el cálculo **dentro de un solo grupo**.

```python
def correlacion_parcial(x, y, z):
    """Correlación entre x e y una vez removido el efecto lineal de z."""
    res_x = x - np.polyval(np.polyfit(z, x, 1), z)
    res_y = y - np.polyval(np.polyfit(z, y, 1), z)
    return np.corrcoef(res_x, res_y)[0, 1]

# 2.3 - ANALISIS DEL PAR MAS CORRELACIONADO

x = df['temp']
y = df['atemp']

# 1. Diagrama de dispersion
plt.figure(figsize=(7, 5))
plt.scatter(x, y, alpha=0.5)
plt.xlabel('Temperatura')
plt.ylabel('Sensacion termica')
plt.title('Relacion entre temperatura y sensacion termica')
plt.grid(alpha=0.3)
plt.show()

print('Correlacion original:', round(x.corr(y), 3))

# 2. Revisar escalas
print('Rango temperatura:', x.min(), x.max())
print('Rango sensacion termica:', y.min(), y.max())

# 3. Correlacion parcial controlando humedad
z = df['hum']

print('Correlacion parcial controlando humedad:',
      round(correlacion_parcial(x, y, z), 3))

# 4. Correlacion dentro de una sola estacion
grupo = df[df['season'] == 3]

print('Correlacion dentro de otoño:',
      round(grupo['temp'].corr(grupo['atemp']), 3))
```

_Tu conclusión sobre ese par. ¿Hay relación real, o hay un confundidor? ¿Cuál?_   

Existe una relación positiva muy fuerte entre temperatura y sensación térmica (0,992). Al controlar la humedad, la correlación no cambia, por lo que no se identifica evidencia de que esta variable sea un confundidor. Dentro de otoño la relación sigue siendo fuerte (0,877). Se concluye que existe una asociación consistente, aunque correlación no significa causalidad.
