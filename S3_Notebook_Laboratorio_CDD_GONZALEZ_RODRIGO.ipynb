### ✏️ Actividad 2.2 — Matriz de correlación

Calcula y grafica la matriz de correlación de tus variables numéricas. Identifica los dos pares con mayor correlación.

```python
corr = df.select_dtypes(include='number').corr()

fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(corr, cmap='RdBu_r', vmin=-1, vmax=1)

ax.set_xticks(range(len(corr)))
ax.set_xticklabels(corr.columns, rotation=90)
ax.set_yticks(range(len(corr)))
ax.set_yticklabels(corr.columns)

for i in range(len(corr)):
    for j in range(len(corr)):
        ax.text(j, i, f'{corr.iloc[i, j]:.2f}',
                ha='center', va='center', fontsize=8)

plt.colorbar(im)
plt.title('Matriz de correlación')
plt.tight_layout()
plt.show()
```

**Conclusión:** las variables presentan relaciones importantes, pero una correlación alta no significa necesariamente que una variable cause cambios en la otra.


### ✏️ Actividad 2.3 — El par más correlacionado, bajo sospecha

Toma el par con mayor correlación y **no te quedes con el número**. Haz el recorrido del Ejercicio 4:

1. Grafica el diagrama de dispersión. ¿La relación es lineal, o el coeficiente engaña?
2. ¿Alguna de las dos variables depende del **tamaño** de la unidad de análisis? Si es así, **normalízalas** y vuelve a calcular.
3. Piensa qué **tercera variable** podría estar causando ambas, y calcula la correlación parcial.
4. Si tienes una variable de agrupación, repite el cálculo **dentro de un solo grupo**.

```python
def correlacion_parcial(x, y, z):
    """Correlación entre x e y una vez removido el efecto lineal de z."""
    res_x = x - np.polyval(np.polyfit(z, x, 1), z)
    res_y = y - np.polyval(np.polyfit(z, y, 1), z)
    return np.corrcoef(res_x, res_y)[0, 1]

# 2.3 - ANALISIS DEL PAR MAS CORRELACIONADO

x = df['temp']
y = df['atemp']

# 1. Diagrama de dispersion
plt.figure(figsize=(7, 5))
plt.scatter(x, y, alpha=0.5)
plt.xlabel('Temperatura')
plt.ylabel('Sensacion termica')
plt.title('Relacion entre temperatura y sensacion termica')
plt.grid(alpha=0.3)
plt.show()

print('Correlacion original:', round(x.corr(y), 3))

# 2. Revisar escalas
print('Rango temperatura:', x.min(), x.max())
print('Rango sensacion termica:', y.min(), y.max())

# 3. Correlacion parcial controlando humedad
z = df['hum']

print('Correlacion parcial controlando humedad:',
      round(correlacion_parcial(x, y, z), 3))

# 4. Correlacion dentro de una sola estacion
grupo = df[df['season'] == 3]

print('Correlacion dentro de otoño:',
      round(grupo['temp'].corr(grupo['atemp']), 3))
```

_Tu conclusión sobre ese par. ¿Hay relación real, o hay un confundidor? ¿Cuál?_   

_Existe una relación positiva muy fuerte entre temperatura y sensación térmica (0,992). Al controlar la humedad, la correlación no cambia, por lo que no se identifica evidencia de que esta variable sea un confundidor. Dentro de otoño la relación sigue siendo fuerte (0,877). Se concluye que existe una asociación consistente, aunque correlación no significa causalidad._

_En el gráfico se observa un dato alejado del comportamiento general, pero la relación entre ambas variables sigue siendo claramente positiva_

## Parte 3 · Series temporales

**Si tu conjunto tiene una variable de fecha, esta parte es obligatoria.** Si no la tiene, reemplázala por un análisis bivariado adicional y dilo explícitamente aquí.

### ✏️ Actividad 3.1 — Construir la serie

Agrega tu conjunto por período (mes, semana o día, según corresponda) y grafica la evolución de tu métrica principal.

```python
# ACTIVIDAD 3.1 - CONSTRUIR LA SERIE TEMPORAL

# Convertir fecha a formato datetime
df['dteday'] = pd.to_datetime(df['dteday'])

# Construir serie mensual de alquileres
serie = df.set_index('dteday')['cnt'].resample('MS').sum()

# Mostrar resultados
print('Período:', serie.index.min(), 'a', serie.index.max())
print('Cantidad de meses:', len(serie))

# Graficar evolución mensual
plt.figure(figsize=(12, 5))
serie.plot(marker='o')

plt.title('Evolución mensual de alquileres de bicicletas')
plt.xlabel('Fecha')
plt.ylabel('Total de alquileres')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

**Interpretación de la serie temporal**

El gráfico muestra que los alquileres de bicicletas aumentaron durante gran parte del período analizado. También se observan variaciones según los meses del año, con aumentos y disminuciones de la demanda.

Durante 2012 se registró una mayor cantidad de alquileres que en 2011.

**Conclusión:** la demanda presenta un crecimiento general y variaciones relacionadas con la época del año.


### ✏️ Actividad 3.2 — Descomposición

Si tu serie tiene al menos **dos ciclos completos** (por ejemplo, 24 meses), descompónla en tendencia, estacionalidad y residuo.

Después responde: ¿qué muestra el **residuo**? ¿Hay algún período anómalo?

```python
# ACTIVIDAD 3.2 - DESCOMPOSICIÓN

from statsmodels.tsa.seasonal import seasonal_decompose

desc = seasonal_decompose(
    serie,
    model='additive',
    period=12
)

fig = desc.plot()
fig.set_size_inches(12, 8)
plt.tight_layout()
plt.show()
```

**Interpretación de la descomposición**

La descomposición muestra que la demanda de bicicletas tiene una tendencia creciente y un comportamiento estacional que se repite durante el año. Los alquileres aumentan en los meses centrales y disminuyen hacia el final del año.

```python
# Revisión de los residuos

print(desc.resid.dropna().round(2))
print('\nResumen de residuos:')
print(desc.resid.describe().round(2))
```

_Tu lectura de la tendencia, la estacionalidad y el residuo:_

**Tendencia:** se observa un crecimiento general de los alquileres entre 2011 y 2012.

**Estacionalidad:** existe un patrón anual repetitivo, con mayor demanda durante los meses centrales del año.

**Residuo:** se obtuvieron 12 valores, todos iguales a -388,91, sin variabilidad entre ellos.

**Conclusión:** la demanda de alquileres presenta una tendencia creciente entre 2011 y 2012 y un comportamiento estacional que se repite cada año. Los 12 residuos obtenidos tienen el mismo valor (-388,91), sin variabilidad. Por este motivo, no se identifican diferencias entre los residuos disponibles. Sin embargo, este resultado debe interpretarse con precaución, ya que solamente se cuenta con 24 meses de información.





### ✏️ Actividad 3.3 — La comparación honesta

Si tu serie tiene estacionalidad, calcula la variación entre dos períodos consecutivos y la variación interanual del mismo período. Muestra la diferencia entre ambas lecturas (Ejercicio 6).

```python
# ACTIVIDAD 3.3 - COMPARACION HONESTA

comparacion = pd.DataFrame({
    'Alquileres': serie,
    'Variacion_mensual_%': serie.pct_change() * 100,
    'Variacion_interanual_%': serie.pct_change(12) * 100
})

comparacion = comparacion.round(2)

comparacion
```

**Interpretación**

La variación mensual muestra aumentos y disminuciones en los alquileres durante el año. En cambio, la variación interanual muestra que todos los meses de 2012 superaron a sus equivalentes de 2011, con incrementos entre 41,33 % y 157,44 %.

**Conclusión:** la demanda de bicicletas aumentó considerablemente durante 2012, aunque presentó variaciones mensuales relacionadas con la estacionalidad.

Esto es lo que la semana 4 va a poner a prueba. Una hipótesis de trabajo no es una corazonada: es una afirmación **falsable** que surge de la evidencia que acabas de producir.

### ✏️ Actividad 4.1 — Formula dos hipótesis

Cada una debe indicar:

- **La afirmación**, en una frase.
- **La evidencia exploratoria** que la sugiere (qué gráfico, qué número, en qué celda).
- **Qué la refutaría**: qué tendrías que observar para descartarla.
- **Qué confundidor podría explicarla** sin que la relación sea real.

Si no puedes escribir qué la refutaría, no es una hipótesis: es una opinión.

**Hipótesis 1**

* **Afirmación:** A mayor temperatura, aumenta la demanda de alquileres de bicicletas.

* **Qué la refutaría:** Que al analizar los datos no exista una relación positiva entre ambas variables.
* **Confundidor posible:** La estación del año y las condiciones climáticas.

**Hipótesis 2**

* **Afirmación:** La demanda de alquileres aumentó durante 2012 respecto de 2011.
* **Evidencia:** La comparación interanual muestra aumentos positivos en los doce meses de 2012.
* **Qué la refutaría:** Que al comparar los mismos meses, la demanda de 2012 sea igual o inferior a 2011.
* **Confundidor posible:** cambios en la disponibilidad de bicicletas o en las condiciones del servicio.
